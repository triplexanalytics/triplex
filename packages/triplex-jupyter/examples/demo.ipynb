# Triplex in a notebook

Shows a small synthetic customer table in the Triplex UI, then reads the results back in Python.
Everything runs locally: the engine is in this browser tab, and the data only moves between the tab and the kernel.

```python
import io

import pandas as pd
from triplex_jupyter import TriplexWidget

df = pd.read_csv("../../../tests/fixtures/sample.customers.csv")

w = TriplexWidget(df)
w
```

Synthetic data from Python: the open widget has no synthesis dialog (that is part of Triplex Pro,
a paid extension), but the `triplex` package runs the same engine on the same table.
Install it as described in `packages/triplex-python`, then run the next cell.

```python
try:
    import triplex
except ImportError:
    print("The triplex Python package is not installed; see packages/triplex-python.")
else:
    run = triplex.synthesize(df, rows=1000, seed=1)
    synthetic = run.table.to_pandas()
    print(run.report.get("method"), "->", synthetic.shape)
    display(synthetic.head())
```

The contract the UI inferred is the same `.triplex.yaml` text the `triplex` CLI reads.

```python
print(w.contract_yaml)
```
